```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier, GradientBoostingRegressor
import pandas as pd
import numpy as np
from xgboost import XGBRegressor # Usa gradientes y derivadas de segundo orden.
```

### **Cargamos el dataset**

```python
df = pd.read_csv("Advertising.csv")
df.head()
```

### **Random forest con TV, radio y newspaper**

Uso `TV`, `radio` y `newspaper` para explicar `sales` con 15 árboles de profundidad máxima 3, al menos 4 muestras por hoja y 4 para dividir un nodo.

```python
from sklearn.model_selection import train_test_split

X = df[["TV", "radio", "newspaper"]]
y = df["sales"]

bosque = RandomForestRegressor(n_estimators=15, max_depth=3, min_samples_leaf=4, min_samples_split=4, random_state=26)
bosque.fit(X, y)
```

### **Train-test 30/70**

Entreno con el 30% de los datos y pruebo con el 70% restante para comparar la R cuadrada de los dos grupos.

```python
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.7, random_state=26)

bosque.fit(X_train, y_train)
r2_train = bosque.score(X_train, y_train)
r2_test = bosque.score(X_test, y_test)

pd.DataFrame({
    "Datos": ["Train", "Test"],
    "R cuadrada": [r2_train, r2_test]
})
```

El bosque explica el 95% de la variación de las ventas en train y el 86% en test. La diferencia de unos 9 puntos muestra algo de sobreajuste, aunque el modelo sigue prediciendo bien con datos que no vio.

### **Random forest sin restricciones**

Repito el bosque con los valores que trae por defecto y la misma partición para comparar la R cuadrada con la del modelo restringido.

```python
bosque_libre = RandomForestRegressor(random_state=26)
bosque_libre.fit(X_train, y_train)
r2_train_libre = bosque_libre.score(X_train, y_train)
r2_test_libre = bosque_libre.score(X_test, y_test)

pd.DataFrame({
    "Datos": ["Train", "Test"],
    "R cuadrada": [r2_train_libre, r2_test_libre]
})
```

Sin restricciones la R cuadrada sube a 0.994 en train y 0.937 en test, frente a 0.951 y 0.861 del bosque restringido. Con solo 60 datos de entrenamiento, los árboles más profundos captan mejor la relación y además generalizan mejor.

```python
bosque_libre.get_params()
```

```python
bosque_libre.estimators_
```

```python
bosque_libre.estimators_[0].get_depth()
```

```python
depth = []

for tree in bosque_libre.estimators_:
    depth.append(tree.get_depth())

np.mean(depth)
```

El primer árbol llega a 7 niveles y el promedio de los 100 árboles es de 8.65, casi el triple de la profundidad 3 del bosque restringido. Por eso el bosque libre se ajusta tanto al train.

### **Gradient boosting con TV, radio y newspaper**

Repito el modelo restringido con `GradientBoostingRegressor`, con 15 árboles de profundidad máxima 3, al menos 4 muestras por hoja y 4 para dividir un nodo, y la misma partición 30/70.

```python
boost = GradientBoostingRegressor(n_estimators=15, max_depth=3, min_samples_leaf=4, min_samples_split=4, random_state=26)
boost.fit(X_train, y_train)
r2_train_boost = boost.score(X_train, y_train)
r2_test_boost = boost.score(X_test, y_test)

pd.DataFrame({
    "Datos": ["Train", "Test"],
    "R cuadrada": [r2_train_boost, r2_test_boost]
})
```

El boosting explica el 93% de la variación de las ventas en train y el 83% en test, un poco menos que el bosque restringido (95% y 86%). Con solo 15 árboles y la tasa de aprendizaje de 0.1, cada árbol corrige una parte pequeña del error de los anteriores y el modelo todavía no termina de ajustarse.

### **Gradient boosting sin restricciones**

Repito el boosting con los valores que trae por defecto y la misma partición.

```python
boost_libre = GradientBoostingRegressor(random_state=26)
boost_libre.fit(X_train, y_train)
r2_train_boost_libre = boost_libre.score(X_train, y_train)
r2_test_boost_libre = boost_libre.score(X_test, y_test)

pd.DataFrame({
    "Datos": ["Train", "Test"],
    "R cuadrada": [r2_train_boost_libre, r2_test_boost_libre]
})
```

Sin restricciones la R cuadrada llega a 0.9998 en train y 0.945 en test, frente a 0.994 y 0.937 del bosque libre. El boosting casi reproduce el train y aun así es el que mejor predice los datos que no vio.

```python
boost_libre.get_params()
```

```python
boost_libre.estimators_
```

```python
depth_boost = []

for tree in boost_libre.estimators_:
    depth_boost.append(tree[0].get_depth())

np.mean(depth_boost)
```

Los 100 árboles tienen profundidad 3, porque el boosting deja `max_depth=3` por defecto. A diferencia del bosque, no necesita árboles profundos, ya que cada árbol nuevo se ajusta a los residuos de los anteriores.
