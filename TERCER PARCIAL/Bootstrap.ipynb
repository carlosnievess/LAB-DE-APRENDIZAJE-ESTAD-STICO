### **Librerías**

```python
import pandas as pd
import statsmodels.api as sm
from sklearn.utils import resample
from sklearn.linear_model import LinearRegression
import numpy as np
```

## **REGRESIÓN LINEAL MÚLTIPLE SIMPLE CON STATSMODEL OLS - ADVERTISING**

```python

df = pd.read_csv("Advertising.csv")

X = df[["TV", "radio", "newspaper"]]
y = df["sales"]

X_ = sm.add_constant(X)

model = sm.OLS(y, X_).fit()

print(model.summary())
```

El modelo explica casi el 90% de la variación de las ventas (R² de 0.897). TV (0.0458) y radio (0.1885) salen significativas con p-values de 0.000, mientras que newspaper tiene un coeficiente casi nulo (-0.0010) con un p-value de 0.860, o sea que el gasto en periódico no aporta a las ventas una vez que ya están TV y radio.

### **Resample**

```python
Xb, yb = resample(X, y, random_state=5,) # Se puede usar replace = False es para sin reemplazo.

display(Xb, yb)
```

La muestra nueva tiene las mismas 200 filas que la original, pero se toma con reemplazo, así que algunas observaciones salen repetidas y otras no aparecen.

### **Regresion lineal multiple lineal con Xb, yb**

```python
lr = LinearRegression()
lr.fit(Xb, yb)
display(lr.coef_, lr.intercept_)

```

Con esta muestra los coeficientes cambian poco en TV (0.0468 frente a 0.0458) y en radio (0.1845 frente a 0.1885), pero newspaper pasa de -0.0010 a 0.0071 y el intercepto baja de 2.94 a 2.53.

### **Lo mismo pero con random_state = 6**

```python
Xb, yb = resample(X, y, random_state=6,)

display(Xb, yb)
```

Aquí ya se ve una fila repetida, la 106 sale dos veces entre las primeras cinco, que es justo lo que pasa al muestrear con reemplazo.

```python
lr = LinearRegression()
lr.fit(Xb, yb)
display(lr.coef_, lr.intercept_)
```

Con la semilla 6 me sale TV en 0.0453, radio en 0.1814 y newspaper otra vez negativo (-0.0020), con un intercepto de 3.27, así que cada muestra da coeficientes un poco distintos.

### **Lo mismo pero con random_state = 7**

```python
Xb, yb = resample(X, y, random_state=7,)

display(Xb, yb)

lr = LinearRegression()
lr.fit(Xb, yb)
display(lr.coef_, lr.intercept_)
```

Con la semilla 7 TV (0.0447) y radio (0.1781) siguen cerca del ajuste original, y newspaper vuelve a cambiar de signo (0.0036). Esto confirma que TV y radio son estables y que newspaper solo se mueve alrededor de cero.

### **For para cambiar la semilla hasta 10,000**

```python
beta_TV = []
beta_radio = []
beta_newspaper = []
interceptos = []

for seed in range(10000):
    Xb, yb = resample(X, y, random_state=seed)
    lr = LinearRegression()
    lr.fit(Xb, yb)
    # display(f"Seed: {seed}", lr.coef_, lr.intercept_)
    beta_TV.append(lr.coef_[0])
    beta_radio.append(lr.coef_[1])
    beta_newspaper.append(lr.coef_[2])
    interceptos.append(lr.intercept_)
```

Repito el remuestreo 10,000 veces, cada una con su semilla, y guardo los coeficientes de cada ajuste para ver cómo se distribuyen.

### **Tablas con betas e intercepto**

```python
beta_TV_df = pd.DataFrame(beta_TV)
beta_radio_df = pd.DataFrame(beta_radio)
beta_newspaper_df = pd.DataFrame(beta_newspaper)
interceptos_df = pd.DataFrame(interceptos)

display("Betas TV:", beta_TV_df)
display("Betas Radio:", beta_radio_df)
display("Betas Newspaper:", beta_newspaper_df)
display("Interceptos:", interceptos_df)
```

Cada fila es el coeficiente que salió de una muestra. En TV y radio los valores se quedan en un rango angosto, pero en newspaper hay tanto negativos como positivos.

### **Tabla con la media y desviación muestral del intercepto y de las demás betas**

```python
media_intercepto = np.mean(interceptos)
media_beta_TV = np.mean(beta_TV)
media_beta_radio = np.mean(beta_radio)
media_beta_newspaper = np.mean(beta_newspaper)

desv_intercepto = np.std(interceptos, ddof=1)
desv_beta_TV = np.std(beta_TV, ddof=1)
desv_beta_radio = np.std(beta_radio, ddof=1)
desv_beta_newspaper = np.std(beta_newspaper, ddof=1)

tabla_medias = pd.DataFrame({
    "Intercepto": [media_intercepto],
    "Beta TV": [media_beta_TV],
    "Beta Radio": [media_beta_radio],
    "Beta Newspaper": [media_beta_newspaper],
    "Desv Intercepto": [desv_intercepto],
    "Desv Beta TV": [desv_beta_TV],
    "Desv Beta Radio": [desv_beta_radio],
    "Desv Beta Newspaper": [desv_beta_newspaper]
})

display(tabla_medias)
```

Las medias del bootstrap (2.953, 0.0457, 0.1884 y -0.0009) son prácticamente iguales a los coeficientes de OLS (2.939, 0.0458, 0.1885 y -0.0010). Las desviaciones estándar (0.337, 0.0019, 0.0109 y 0.0065) son la estimación bootstrap del error estándar y quedan cerca de las de OLS (0.312, 0.001, 0.009 y 0.006), aunque un poco más grandes porque el bootstrap no supone que la varianza de los errores sea constante. En newspaper la desviación es varias veces mayor que la media, así que el coeficiente no se distingue de cero, igual que con su p-value.
