## **EJERCICIO CON "bateria_tableta.txt"**

```python
# Leer el archivo "bateria_tableta.txt":

import pandas as pd

df = pd.read_csv("bateria_tableta.txt")

df
```

```python
# Gráfica de datos iniciales

import matplotlib.pyplot as plt

plt.scatter(df['carga'], df['duracion'])
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Gráfica de Duración vs Carga')
plt.show()
```

## MODELO - B0 + B1X

```python
# Paqueterías faltantes

import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# Definimos x y y

x = df[[ 'carga' ]]
y = df[['duracion']]
```

```python
# Modelo

model = LinearRegression()
model.fit(x, y)
```

```python
# Imprimir ecuación

b0 = model.intercept_[0]
b1 = model.coef_[0][0]

print(f"Ecuación del modelo: y = {b0:.4f} + {b1:.4f}x")
```

```python
# Gráfica de Duración vs Carga con la línea de regresión

plt.scatter(x, y, color='blue', label='Datos')
plt.plot(x, model.predict(x), color='red', label='Línea de regresión')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Gráfica de Duración vs Carga con Línea de Regresión')
plt.legend()
plt.show()
```

## MODELO - B0 + B1X + B2X2 + B3X3 + B4X4 + B5X5

```python
# Crear modelo de quinto orden

X_v = np.reshape(x, (-1, 1))

X = np.hstack([
    X_v,
    X_v**2,
    X_v**3,
    X_v**4,
    X_v**5
])

lr = LinearRegression()
lr.fit(X, y)

predicciones = lr.predict(X)

print("B0:", lr.intercept_)
print("B1 a B5:", lr.coef_)
```

```python
# Imprimir la ecuación del modelo de quinto orden

print(f"{lr.intercept_[0]:.4f} + {lr.coef_[0][0]:.4f}x + {lr.coef_[0][1]:.4f}x^2 + {lr.coef_[0][2]:.4f}x^3 + {lr.coef_[0][3]:.4f}x^4 + {lr.coef_[0][4]:.4f}x^5")
```

```python
# Gráfica de Duración vs Carga con la línea de regresión

plt.scatter(x, y, color='blue', label='Datos')
plt.scatter(x, predicciones, color='red', label='Línea de regresión de quinto orden')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Gráfica de Duración vs Carga con Línea de Regresión de Quinto Orden')
plt.legend()
plt.show()
```

## MODELO - B0 + B1X + B2X2 + B3X3 + B4X4 + B5X5 + B6X6 + B7X7 + B8X8 + B9X9

```python
# Crear modelo de noveno orden

X_v = np.reshape(x, (-1, 1))

X = np.hstack([
    X_v,
    X_v**2,
    X_v**3,
    X_v**4,
    X_v**5,
    X_v**6,
    X_v**7,
    X_v**8,
    X_v**9
])

lr = LinearRegression()
lr.fit(X, y)

predicciones = lr.predict(X)

print("B0:", lr.intercept_)
print("B1 a B9:", lr.coef_)
```

```python
# Imprimir la ecuación del modelo de noveno orden

print(f"{lr.intercept_[0]:.4f} + {lr.coef_[0][0]:.4f}x + {lr.coef_[0][1]:.4f}x^2 + {lr.coef_[0][2]:.4f}x^3 + {lr.coef_[0][3]:.4f}x^4 + {lr.coef_[0][4]:.4f}x^5 + {lr.coef_[0][5]:.4f}x^6 + {lr.coef_[0][6]:.4f}x^7 + {lr.coef_[0][7]:.4f}x^8 + {lr.coef_[0][8]:.4f}x^9")
```

```python
# Gráfica de Duración vs Carga con la línea de regresión

plt.scatter(x, y, color='blue', label='Datos')
plt.scatter(x, predicciones, color='red', label='Línea de regresión de noveno orden')
plt.xlabel('Carga')
plt.ylabel('Duración')
plt.title('Regresión de Noveno Orden')
plt.legend()
plt.show()
```

## **EJERCICIO CON "Salarios_minimos.csv"**

```python
# Leer el archivo "Salarios_minimos.csv":

df = pd.read_csv("Salarios_minimos.csv")

df
```

```python
# Definimos variables

x = df['Periodo']
y = df[['Salario']]

# Creación de gráfica

plt.scatter(x, y)
plt.xlabel('Año')
plt.ylabel('Salario')
plt.title('Salario Mínimo vs Año')
plt.show()

```

## MODELO - B0 + B1X

```python
# Modelo de primer orden

X_v = np.reshape(x, (-1, 1))

X = np.hstack([
    X_v,])

lr = LinearRegression()
lr.fit(X_v, y)

predicciones = lr.predict(X)

print("B0:", lr.intercept_)
print("B1:", lr.coef_)
```

```python
# Ecuación del modelo de primer orden

print(f"{lr.intercept_[0]:.4f} + {lr.coef_[0][0]:.4f}x")
```

```python
# Gráfica de Salario Mínimo vs Año con la línea de regresión

plt.scatter(x, y, color='blue', label='Datos')
plt.plot(x, predicciones, color='red', label='Línea de regresión de primer orden')
plt.xlabel('Año')
plt.ylabel('Salario')
plt.title('Salario Mínimo vs Año')
plt.legend()
plt.show()
```

```python
# Ecuación

print(f"{lr.intercept_[0]:.4f} + {lr.coef_[0][0]:.4f}x + {lr.coef_[0][1]:.4f}x^2 + {lr.coef_[0][2]:.4f}x^3 + {lr.coef_[0][3]:.4f}x^4 + {lr.coef_[0][4]:.4f}x^5")
```
